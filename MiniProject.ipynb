# 4.1 Comparing Study Hours and Assessment Scores
Claim: If students study 3 hours or more per day (Group 1), than they receive a greater assessment score than students who study less than 3 hours per day (Group 2). 

H0 (Null Hypothesis): µa (Group1) = µb (Group2)

H0: There is no significant difference in Group 1 and Group 2 Assessment scores. 

H1 (Alternative Hypothesis): µa (Group1) > µb (Group2)

H1: There is a significantly greater difference in Group 1 assessment score than in Group 2. 


In this section i will use the structure and my knowldge from Weeks 9 & 10 lectures on hypothesis testing to test my claim. I start by loading the study data csv and seperating the study hours into two groups. Group 1 contains study hours of 3 or more and Group 2 contains study hours of less than 3 hours. As assessment scores and study hours are independent observations, this is a unpaired setting. In this case we will be using only independent tests. I then used a box plot to visualise the two groups distribution. A Shapiro Wilk test was done to determine whether the distribution is normal or not. The Shapiro normality test for Group 1 shows a p-value of 0.47 and a test statistics of 0.95. The normality test for Group 2 shows a p-value of 0.69 and a test statistics of 0.95. The Levene test was used to find equal variance between Group 1 and Group 2. The p-value shows 0.64 and the test statistics shows 0.21. As the results for the normality test showed a normal distribution a parametric student t test (independent test) was conducted, which found a p value of 5.33e-07 (which calculates to 0.000000533)and a test statistics of 6.20. A 95% confidence interval was done on the mean assessment score which resulted in 65.5 and 75.4. 

```python
studydata = pd.read_csv('StudyData.csv')

df_group1 = studydata[(studydata['StudyHours']=='3 hours') | (studydata['StudyHours']=='4 hours')]
df_group2 = studydata[(studydata['StudyHours']=='Less than 1 hour') | (studydata['StudyHours']=='2 hours')]

df_group1.head()
df_group2.head()

x1 = df_group1['AssessmentScores'].to_numpy()
x2 = df_group2['AssessmentScores'].to_numpy()

fig,ax=plt.subplots(1, figsize=(5,4))
sp.stats.plot_groups_boxes([x1,x2], lines=False,
    xlabel=['Group 1 (3+ hours)', 'Group 2 (Less than 3 hours)'],ylab='Assessment Score', title='Two Groups: Unpaired', ax=ax)
```

```python
def QQplot(x):
    fig, ax = plt.subplots(1, figsize=(4,3))
    data_sorted=np.sort(x)
    n = len(data_sorted)
    p = (np.arange(1, n+1) - 0.5) / n
    theoretical = scipy.stats.norm.ppf(p)*data_sorted.std()+data_sorted.mean()
    plt.scatter(theoretical, data_sorted)
    plt.plot(theoretical, theoretical, color='C3')  
    plt.xlabel("Theoretical quantiles")
    plt.ylabel("Sample quantiles")
    plt.title('QQ-Plot')
    plt.grid(alpha=0.5)
    plt.show()
```

```python
t, p = scipy.stats.shapiro(x1)
print('Group 1 test statistics :', t)
print('Group 1 p-value : \t', p)
QQplot(x1)

t, p = scipy.stats.shapiro(x2)
print('Group 2 test statistics :', t)
print('Group 2 p-value :\t', p)
QQplot(x2)
 
t, p = scipy.stats.levene(x1, x2)
print('test statistics :', t)
print ('p-value : \t', p)

t, p = scipy.stats.ttest_ind(x1, x2, equal_var=True, alternative='greater')
print('test statistics :', t)
print('p-value:\t', p)

CI95 = (population_mean_est - 1.96*se, population_mean_est + 1.96*se)
CI95
```

# 4.2 Comparing Attendance and Assessment Scores
Claim: If students have a high attendence (Group 1), than they receive a greater assessment score than students with low attendance (Group 2).

H0 (Null Hypothesis): µ(att1) = µ(att2)

H0: There is no significant difference in the assessment scores of students in Group 1 and Group 2.

H1 (Alternative Hypothesis): µ(att1) ≠ µ(att2)

H1: There is a significant difference in the assessment scores of students in Group 1 and Group 2.

In this section i will compare high and low attendance scores against students assessment scores. I start by loading the data in the study data csv and sperating them into two groups. I used a Likert scale of 1 to 5 to measure the values of students attendance. Group 1 represents high attendance (4 and 5) and Group 2 represents low attendance (1, 2, 3). Both variables, attendance and assessment score, are independent observations therefore this is a unpaired setting and independent tests will be done. The QQ box plot is used to visualise the distribution in low attendance and high attendance. The Shapiro Wilk normality test results in a p-value of 0.45 and a test statistics of 0.95 in Group 1. In Group 2 the p-value is 0.55 and the test statistics is 0.95. This test and the box plot both indicate normal distribution, therefore we will comtinue with a parametric test. This also confirmed with the Levene equal variance test, which shows a p-value of 0.79 and a test statistics of 0.07. The parametric test used it the Student t test(independent), which presents a p-value of 4.58e-06 (which equals to 0.00000458) and a test statistics of 5.65.

```python
df_att1 = studydata[(studydata['Attendance']==5) | (studydata['Attendance']== 4)]
df_att2 = studydata[(studydata['Attendance']==1) | (studydata['Attendance']== 2) | (studydata['Attendance']== 3)]

x1_att = df_att1['AssessmentScores'].to_numpy()
x2_att = df_att2['AssessmentScores'].to_numpy()

fig, ax = plt.subplots(1, figsize=(5,4))
sp.stats.plot_groups_boxes([x1_att, x2_att], lines=False,
                          xlabels=['High Attendance', 'Low Attendance'],
                          ylab='Assessment Score',
                          title='Unpaired Groups', ax=ax)

t, p = scipy.stats.shapiro(x1_att)
print('Group 1 (High Attendance) test statistics :', t)
print('Group 1 (High Attendance) p-value : \t', p)
QQplot(x1_att)

t, p = scipy.stats.shapiro(x2_att)
print('Group 2 (Low Attendance) test statistics :', t)
print('Group 2 (Low Attendance) p-value :\t', p)
QQplot(x2_att)

t, p = scipy.stats.levene(x1_att, x2_att)
print('test statistics :', t)
print('p-value :\t', p)

t, p =scipy.stats.ttest_ind(x1_att, x2_att, equal_var=True, alternative='two-sided')
print('tests statistics :', t)
print('p-value :\t', p)
```

# 4.4 Correlation between Study Hours and Assessment Score

```python
study_hours_map = {'Less than 1 hour': 1, '2 hours': 2, '3 hours': 3, '4 hours': 4}
studydata['StudyHours_num'] = studydata['StudyHours'].map(study_hours_map)

x = studydata['StudyHours_num'].to_numpy()
y = studydata['AssessmentScores'].to_numpy()

r, p = scipy.stats.pearsonr(x,y)
print('Correlation coefficient (r) :', r)
print('p-value : \t', p)
```

# 7.1 Correlation between Sleep Hours and Study Hours
In this section I will perform two correlation tests using exploratory data however staying within the original study. In Week 11 Bajaj (2026) defines a correlation test as an association of two variables for example the association change in x with change in y. In his book on why we sleep, Walker (2017) states that sleep has cognitive benefits, aids our memory and sleeping six hours or less a day reduces learning restoration benefits that usually occur during sleep spindles. Before the conducting the test I clean the data by removing spaces in the study data csv in column sleep hours. This is because python produces errors that come from invisible spaces in the data. Next I encode the data to numerals as correlation tests only use numerical values. 4 to 5 hours was changed to 4.5, 5 to 6 hours becomes 5.5, 6 to 7 hours becomes 6.5 and 8 hours or more is 8, this way I can preserve the order of the data. Next, I load both study hours and sleep hours data and present a scatter graph for visual representation. I will use the non parametric spearman correlation test which produced a p -value of 0.37 and r (correlation coefficient) of 0.16. 

The week 11 interpretation of a correlation test is (r < 0.4) as the coefficients results is 0.16 this suggests a weak positive correlation. However, the p-value result is greater than 0.05 therefore we fail to reject the null hypothesis. The results of this test tells us there isn’t a significant correlation between study hours and sleep. In Week 9 (Bajaj) states that a meaningful relevance does not always come from statistically significant findings. This correlation test was interesting and a common variable in academic performance. Hassan (2025) conducted a similar study testing the association between academic performance and sleep hours in undergraduate medical students. The study found a sleep to be a negative effect of students well being and academic performance, with 73.7% of participants recorded to have poor sleep quality. 

```python
studydata['SleepHours'] = studydata['SleepHours'].str.strip()
studydata['SleepHours'] = studydata['SleepHours'].str.replace('  ', ' ')

sleep_map = {'4 to 5 hours': 4.5, '5 to 6 hours': 5.5, '6 to 7 hours': 6.5, '8 hours or more': 8}
studydata['SleepHours_num']= studydata['SleepHours']. map(sleep_map)

x_sleep = studydata['SleepHours_num']. to_numpy()
y_studyhours = studydata['StudyHours_num'].to_numpy()

fig, ax = plt.subplots(1, figsize=(5,4))
plt.scatter(x_sleep, y_studyhours)
plt.xlabel('Sleep Hours')
plt.ylabel('Study Hours')
plt.title('Sleep Hours and Study Hours')
plt.grid(alpha=0.5)
plt.show()

r, p = scipy.stats.spearmanr(x_sleep, y_studyhours)
print('Correlation coefficient (r) :', r)
print('p-value :\t', p)
```

# 7.2 Correlation between Social Media and Assessment Scores

In this test I considered whether the time spent on social media has an associations with assessment scores. As social media is an ordinal variable I will be using the non parametric spearman correlation test. Macit et al (2018) found in their study that people who stay on social media for several hours develop psychological behaviours such as addiction as well as symptoms such as bipolar disorder and depression. The aim of the exploratory is to consider social media usage and high dopamine intake from is impacts students assessment scores. Similarly, to the first correlation test, I performed a non parametric spearman test as the values for social media are ordinal. 

I then encoded the hours spent on social media into numerical values for instance less than 1 hour is 0.5, 1 to 2 hours is 1.5, 3 to 4 hours is 3.5 and 5 to 6 hours is 5.5. Next I load the data and use numpy to assign the y label to assessment scores and the x label to social media. The scatter graphs is used to show the distribution of social media hours. The spearman correlation test presents a p-value of 8.36 (which results to 0.00000836) and a r value of -0.71. Using the lecture interpretations (r > 0.7) we can assume a strong negative correlation. The correlation coefficient suggested that when assessment scores decreases then social media hours would increase. As the p-value is significantly lower than 0.001, the null hypothesis is rejected and the alternative test is favoured. 

```python
studydata['SocialMedia'] = studydata['SocialMedia'].str.strip()

social_map = {'Less than 1 hour': 0.5, '1 to 2 hours': 1.5, '3 to 4 hours': 3.5, '5 to 6 hours': 5.5 }

studydata['SocialMedia_num']= studydata['SocialMedia']. map(social_map)

x_social = studydata['SocialMedia_num'].to_numpy()
y_scores = studydata['AssessmentScores'].to_numpy()

fig, ax = plt.subplots(1, figsize=(5,4))
plt.scatter(x_social, y_scores)
plt.xlabel('Social Media Hours')
plt.ylabel('Assessment Score')
plt.title('Social Media and Assessment Score')
plt.grid(alpha=0.5)
plt.show()

r, p = scipy.stats.spearmanr(x_social, y_scores)
print('Correlation coefficient (r) :', r)
print('p-value :\t', p)
```
